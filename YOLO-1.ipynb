Vehicle Detection and Tracking

```python
import cv2
```

```python
from ultralytics import YOLO
```

```python
from collections import defaultdict
```

Load the YOLO model

```python
model= YOLO("best.pt")
```

```python
class_list= model.names
class_list
```

Open the video file

```python
cap= cv2.VideoCapture(r"L:\CCTV\20250616\N02174326.mp4")
```

Run the Object Detection by Tracking

```python
# Break the loop when frame is not read.
while cap.isOpened():
    ret, frame= cap.read()
    if not ret:
        break

# Run YOLO tracking on the frame
    results= model.track(frame, persist=True)
    print(results)

# Ensure results are not empty
    if results[0].boxes.data is not None:

# Get the detected boxes, their class indices, and track IDs
        boxes = results[0].boxes.xyxy.cpu()
        track_ids = results[0].boxes.id.int().cpu().tolist()
        class_indices = results[0].boxes.cls.int().cpu().tolist()
        confidences = results[0].boxes.conf.cpu()
        
# For red line
        cv2.line(frame, (690, 430), (1130, 430), (0, 0, 255), 3)
        
# Loop through each detected object
    for box, track_id, class_idx, conf in zip(boxes, track_ids, class_indices, confidences):
        x1, y1, x2, y2 = map(int, box)

# Calculate the center of the bounding box
        cx= (x1 + x2) // 2
        cy= (y1 + y2) // 2

        class_name = class_list[class_idx]

        cv2.circle(frame, (cx, cy), 4, (0, 0, 255), -1)
        cv2.putText(frame, f"ID: {track_id} {class_name}", (x1, y1 - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 255), 2)
        cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)   

#  Show the frame   
    cv2.imshow("YOLO Object Tracking and Counting", frame)
    
# Exit loop if "Q" is pressed
    if cv2.waitKey(1) & 0xFF== ord("q"):
        break

# Release Resources
cap.release()
cv2.destroyAllWindows()
```
